```python
def country_to_continent(country_name):
  if not isinstance(country_name, str):
    return None
  try:
    country_alpha2 = mac.country_name_to_country_alpha2(country_name)
    country_continent_code = mac.country_alpha2_to_continent_code(country_alpha2)
    country_continent_name = mac.convert_continent_code_to_continent_name(country_continent_code)
    return country_continent_name
  except:
      return None
```

```python
internet_data = pd.read_csv("https://drive.google.com/uc?export=view&id=1gGNAjFY_PZnirvI4YBAEhhF99TpYowjt", header=1)
internet_data["Continent"] = internet_data["Unnamed: 1"].apply(country_to_continent)

internet_croatia = internet_data[ internet_data["Unnamed: 1"] == "Croatia"]
internet_slovakia = internet_data[ internet_data["Unnamed: 1"] == "Slovakia"]
internet_europe = (internet_data[ internet_data["Continent"] == "Europe"].groupby("Year", as_index=False)["Value"].mean())

plt.plot(internet_slovakia['Year'], internet_slovakia['Value'], label ='Slovakia', color='royalblue')
plt.plot(internet_croatia['Year'], internet_croatia['Value'], label ='Croatia', color='palegreen')
plt.plot(internet_europe['Year'], internet_europe['Value'], label ='Europe', color='peru')

plt.xlabel("Year")
plt.ylabel("Internet connectivity (% of all population)")
plt.ylim((0,100))
plt.legend()
plt.title('multiple plots')
plt.show()
```
